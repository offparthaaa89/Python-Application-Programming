```python
# 1. Create an empty dictionary

inventory = {}
```

```python
# 2. Now store the first product's details in variables
ProductNo1_Name = "Mobile Phone"
ProductNo1_Quantity = 5
ProductNo1_Price = 20000
ProductNo1_releaseYear = 2020
```

```python
# 3. Now add the first product's details into inventory
inventory["ProductNo1_Name"] = ProductNo1_Name
inventory["ProductNo1_Quantity"] = ProductNo1_Quantity
inventory["ProductNo1_Price"] = ProductNo1_Price
inventory["ProductNo1_releaseYear"] = ProductNo1_releaseYear
```

```python
# 4. Its time to store the second product's details in variables
ProductNo2_Name = "Laptop"
ProductNo2_Quantity = 3
ProductNo2_Price = 55000
ProductNo2_releaseYear = 2022
```

```python
# 5. This time add the second product's details into inventory
inventory["ProductNo2_Name"] = ProductNo2_Name
inventory["ProductNo2_Quantity"] = ProductNo2_Quantity
inventory["ProductNo2_Price"] = ProductNo2_Price
inventory["ProductNo2_releaseYear"] = ProductNo2_releaseYear
```

```python
# 6. Now display the products present in inventory
print("Inventory contents:")
for key, value in inventory.items():
    print(f"  {key}: {value}")
```

```python
# 7. Check if the release years are present as keys in inventory
print("\nChecking for release year keys:")
print("  'ProductNo1_releaseYear' in inventory:", "ProductNo1_releaseYear" in inventory)
print("  'ProductNo2_releaseYear' in inventory:", "ProductNo2_releaseYear" in inventory)
 
```

```python
# 8. Delete release year of both products from inventory
del inventory["ProductNo1_releaseYear"]
del inventory["ProductNo2_releaseYear"]
 
print("\nInventory after deleting both release years:")
for key, value in inventory.items():
    print(f"  {key}: {value}")
```

```python
print("\n" + "=" * 50)
print("PART B - TUPLES")
print("=" * 50)
 
# Create a tuple of product prices
prices = (250, 300, 150, 400, 100, 350, 200)
print("\nOriginal tuple:", prices)
 
# Find and print the highest and lowest price
highest_price = max(prices)
lowest_price = min(prices)
print("Highest price:", highest_price)
print("Lowest price:", lowest_price)
 
# Calculate and print the total sum of all prices
total_sum = sum(prices)
print("Total sum of prices:", total_sum)
 
# Convert the tuple into a sorted list (ascending order)
sorted_prices = sorted(prices)
print("Sorted prices (ascending, as a list):", sorted_prices)
 
# Try to modify an element in the tuple
print("\nAttempting to modify prices[0] = 500 ...")
try:
    prices[0] = 500
except TypeError as e:
    print("Error occurred:", e)
    print("Explanation: Tuples are immutable in Python, meaning once created,")
    print("their elements cannot be changed, added, or removed. This is what")
    print("distinguishes a tuple from a list. To 'change' a tuple, you must")
    print("create a new tuple (or convert it to a list first).")
```
