```python
# Integer
age = 21
print("Integer:", age, type(age))

# Float
height = 5.8
print("Float:", height, type(height))

# String
name = "Riya"
print("String:", name, type(name))

# Boolean
is_student = True
print("Boolean:", is_student, type(is_student))
```

```python
# Float to Integer
f_value = 9.75
int_from_float = int(f_value)
print("Float to Integer:", int_from_float)

# String to Integer
s_value = "100"
int_from_string = int(s_value)
print("String to Integer:", int_from_string)

# Boolean to Integer
b_value = True
int_from_bool = int(b_value)
print("Boolean to Integer:", int_from_bool)   # True -> 1

b_value2 = False
int_from_bool2 = int(b_value2)
print("Boolean to Integer:", int_from_bool2)  # False -> 0
```

```python
a = 20
b = 4

# Addition
print("Addition:", a + b)

# Subtraction
print("Subtraction:", a - b)

# Multiplication
print("Multiplication:", a * b)

# Division
print("Division:", a / b)
```
