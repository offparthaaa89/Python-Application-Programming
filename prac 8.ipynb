```python
sales = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Aman", "Riya", "Karan", "Sneha", "Rahul",
             "Priya", "Vikas", "Anjali", "Mohit", "Pooja"],
    "Age": [25, 34, np.nan, 29, 41, 38, 22, np.nan, 45, 31],
    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],
    "Purchase_Amount": [1200, 2500, 1800, np.nan, 3200,
                        2100, 900, 2750, 1500, 2300],
})
 
print("\nProblem 4: first 5 rows")
print(sales.head())
 
print("\nSummary statistics")
print(sales.describe())
 
print("\nMissing values before handling")
print(sales.isnull().sum())
 
# Handling: fill numeric columns with median (robust to outliers)
sales["Age"] = sales["Age"].fillna(sales["Age"].median())
sales["Purchase_Amount"] = sales["Purchase_Amount"].fillna(sales["Purchase_Amount"].median())
 
print("\nMissing values after handling")
print(sales.isnull().sum())
```

```python
print("\nProblem 5a: Customers with Age > 30")
print(sales[sales["Age"] > 30])
 
print("\nProblem 5b: Total Purchase_Amount by Gender")
print(sales.groupby("Gender")["Purchase_Amount"].sum())
 
print("\nProblem 5c: Customer_ID, Name, Purchase_Amount")
print(sales[["Customer_ID", "Name", "Purchase_Amount"]])
```

```python
sales["Discounted_Price"] = sales["Purchase_Amount"] * 0.90
print("\nProblem 6a: With Discounted_Price")
print(sales)
 
print("\nProblem 6b: Average Purchase_Amount by Gender")
print(sales.groupby("Gender")["Purchase_Amount"].mean())
 
print("\nProblem 6c: Sorted by Purchase_Amount (descending)")
print(sales.sort_values(by="Purchase_Amount", ascending=False))
```
