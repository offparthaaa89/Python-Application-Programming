```python
# create an empty list first
shopping_list = []
print("Empty list created:", shopping_list)
```

```python
# Now store the itmes in the shopping list 
shopping_list = ["Watch", "Laptop", "Shoes", "Pen", "Clothes"]
print("Shopping list:", shopping_list)
print("Number of items:", len(shopping_list))
```

```python
# now add football
shopping_list.append("Football")
print("List after adding Football:", shopping_list)
```

```python
# Now print the last item
print("Last item:", shopping_list[-1])
```

```python
# print the entire list
print("Entire shopping list:", shopping_list)
```

```python
# Print the important items 
important_items = shopping_list[1:3]
print("Important items:", important_items)
```

```python
# Now replace the pen with nootebook
index_of_pen = shopping_list.index("Pen")
shopping_list[index_of_pen] = "Notebook"
print("List after replacing Pen with Notebook:", shopping_list)
```

```python
# now delete the least items from the list and print it..
shopping_list.remove("Clothes")
print("Final shopping list:", shopping_list)
```
