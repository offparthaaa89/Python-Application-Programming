```python
class Person:
    def __init__(self, name, id):
        self.name = name
        self.id = id

    def Display(self):
        print(self.name, self.id)


class Emp(Person):
    def __init__(self, name, id):
        super().__init__(name, id)

    def Print(self):
        print("Emp class called")


Emp_details = Emp("lalit", 103)

Emp_details.Display()
Emp_details.Print()
```

```python
word1 = input("Enter first word: ")
word2 = input("Enter second word: ")

if sorted(word1) == sorted(word2):
    print("Anagram")
else:
    print("Not Anagram")
```

```python
num = int(input("Enter a number: "))

original = num
digits = len(str(num))
sum = 0

while num > 0:
    digit = num % 10
    sum = sum + digit ** digits
    num = num // 10

if sum == original:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
class Vehicle:

    color = "white"

    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed
        self.mileage = mileage
        self.seating_capacity = 0

    def assign_seating_capacity(self, capacity):
        self.seating_capacity = capacity

    def display(self):
        print("Maximum Speed:", self.max_speed, "kmph")
        print("Mileage:", self.mileage, "kmpl")
        print("Color:", self.color)
        print("Seating Capacity:", self.seating_capacity)


car1 = Vehicle(200, 20)
car1.assign_seating_capacity(5)

car2 = Vehicle(180, 25)
car2.assign_seating_capacity(4)

print("Car 1")
car1.display()

print()

print("Car 2")
car2.display()
```
